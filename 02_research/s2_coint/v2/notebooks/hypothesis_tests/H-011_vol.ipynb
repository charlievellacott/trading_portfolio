# H-011 · Vol target

Book vol target on research IS only. Do not maximise the holdout. Freeze `VOL_STAR`.

After typing STAR, open the sealed holdout **once** (v2 OOS tearsheet) — not in a loop.

Next: open the sealed holdout once (v2 OOS tearsheet).


## 0. Imports & Config


```python
import os
import sys

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

ROOT = os.path.abspath(os.getcwd())
while not os.path.isdir(os.path.join(ROOT, "01_data", "ingestion")):
    parent = os.path.dirname(ROOT)
    if parent == ROOT:
        break
    ROOT = parent
if ROOT not in sys.path:
    sys.path.insert(0, ROOT)

from backtest.s2_coint.report import (
    arm_selection_table,
    fold_table,
    fold_val_metrics,
    full_is_metrics,
    load_star_stack,
    median_sharpe_hint,
    plot_fold_boxplots,
    require_star,
    save_star_stack,
    write_tearsheet_pdf,
)
from backtest.s2_coint.research import (
    config_from_stack,
    is_end_for_stack,
    load_s1_weekly,
    load_s2_cuts,
    load_universe_panels,
    lookbacks_for_bar,
    n_trials_local,
    n_trials_stack,
    overlay_ols_hedge,
    overlay_kalman_hedge,
    register_hypothesis_arms,
    s2_artifacts_dir,
    s2_data_dir,
    tearsheet_path,
)
from backtest.s2_coint.runner import run_s2_backtest
from backtest.s2_coint.walkforward import embargo_bars_for_config, make_s2_folds

VERSION = "v2"
STAR_PATH = os.path.join(s2_artifacts_dir(version=VERSION, root=ROOT), "s2_star_stack.json")
LEDGER_PATH = os.path.join(os.path.dirname(STAR_PATH), "s2_variant_ledger.json")
cuts = load_s2_cuts(root=ROOT, version=VERSION)
stack = load_star_stack(STAR_PATH)
print("yahoo_drop_rate", cuts.get("yahoo_drop_rate"))
print("stack keys", sorted(stack.keys()))

HYP_ID = "H-011"
VOL_ARMS = ("fixed_k", "s1_vt")
require_star("HEDGE_STAR", stack.get("HEDGE_STAR"))
require_star("OVERLAP_STAR", stack.get("OVERLAP_STAR"))
require_star("HL_GATE_STAR", stack.get("HL_GATE_STAR"))

require_star("POOL_STAR", stack.get("POOL_STAR"))
require_star("FDR_STAR", stack.get("FDR_STAR"))
require_star("BOOK_STAR", stack.get("BOOK_STAR"))
require_star("N_STAR", stack.get("N_STAR"))
PAIRS_STAR = list(stack.get("PAIRS_STAR") or [])
if not PAIRS_STAR:
    raise ValueError("PAIRS_STAR empty")
BAR = str(stack.get("BAR_STAR") or "1d")
train, full = load_universe_panels(None, BAR, PAIRS_STAR, root=ROOT, version=VERSION)
lb0 = lookbacks_for_bar(
    BAR,
    ols_days=stack.get("OLS_WINDOW_STAR"),
    z_days=stack.get("Z_WINDOW_STAR"),
    adf_days=stack.get("ADF_WINDOW_STAR"),
)
train = overlay_ols_hedge(train, ols_window=lb0["ols_window"], z_window=lb0["z_window"],
                          hl_window=lb0["hl_window"], adf_window=lb0["adf_window"])
full = overlay_ols_hedge(full, ols_window=lb0["ols_window"], z_window=lb0["z_window"],
                         hl_window=lb0["hl_window"], adf_window=lb0["adf_window"])
is_end = is_end_for_stack(stack, full if BAR == "1h" else train)
if BAR == "1d":
    is_panel, oos_panel = train.copy(), full.loc[pd.to_datetime(full["date"]) > is_end].copy()
else:
    from backtest.s2_coint.research import split_is_oos
    is_panel, oos_panel = split_is_oos(full, is_end=is_end)
s1_weekly = load_s1_weekly(ROOT)
print("bar", BAR, "is_end", is_end, "IS", len(is_panel), "pairs", len(PAIRS_STAR))

```

## 1. Walk-forward


```python
panel = is_panel.copy()
dates = pd.DatetimeIndex(pd.to_datetime(panel["date"])).sort_values().unique()
folds = make_s2_folds(dates, n_folds=3, embargo_bars=embargo_bars_for_config(bar=BAR))
display(fold_table(folds))
configs = {arm: config_from_stack(stack, vol_mode=arm) for arm in VOL_ARMS}
register_hypothesis_arms(HYP_ID, list(configs.keys()), overwrite=True, path=LEDGER_PATH)
fold_df = fold_val_metrics(panel, folds, configs, hyp_id=HYP_ID, s1_weekly=s1_weekly)
full_is_df = full_is_metrics(panel, configs, hyp_id=HYP_ID, s1_weekly=s1_weekly)
plot_fold_boxplots(fold_df, title="H-011 vol fold-val")
plt.show()
print("median-Sharpe hint (commentary only):", median_sharpe_hint(fold_df))
display(arm_selection_table(fold_df, full_is_df))
```

## 2. Type `VOL_STAR`


```python
VOL_STAR = ""  # type: "fixed_k" | "s1_vt"
require_star("VOL_STAR", VOL_STAR)
stack["VOL_STAR"] = VOL_STAR
if VOL_STAR == "s1_vt" and stack.get("VT_TARGET_ANN_VOL_STAR") is None:
    stack["VT_TARGET_ANN_VOL_STAR"] = 0.06  # choose on research IS only; do not maximise holdout
save_star_stack(STAR_PATH, stack)
print("wrote", STAR_PATH, "VOL_STAR", VOL_STAR)
```

## 3. Sealed holdout once (after the full stack is frozen)

Run only after H-011 STAR is typed. Kill if holdout net Sharpe < 0.3 or >40% of PnL from one pair.
Do not retune on this slice.

Next: open the sealed holdout once (v2 OOS tearsheet).


```python
# Optional inline sealed look — prefer a dedicated v2 OOS tearsheet notebook.
RUN_SEALED = False  # set True once, after the stack is frozen
if RUN_SEALED:
    stack = load_star_stack(STAR_PATH)
    oos_feat = oos_panel.copy()
    lb = lookbacks_for_bar(BAR, ols_days=stack.get("OLS_WINDOW_STAR"), z_days=stack.get("Z_WINDOW_STAR"),
                           adf_days=stack.get("ADF_WINDOW_STAR"))
    if str(stack.get("HEDGE_STAR")) == "kalman":
        oos_feat = overlay_kalman_hedge(oos_feat, z_window=lb["z_window"], hl_window=lb["hl_window"],
                                        adf_window=lb["adf_window"],
                                        delta=float(stack.get("KALMAN_DELTA_STAR") or 1e-7))
    else:
        oos_feat = overlay_ols_hedge(oos_feat, ols_window=lb["ols_window"], z_window=lb["z_window"],
                                     hl_window=lb["hl_window"], adf_window=lb["adf_window"])
    cfg = config_from_stack(stack)
    n_loc = n_trials_local(configs)
    n_stk = n_trials_stack(HYP_ID, configs, path=LEDGER_PATH)
    oos = run_s2_backtest(oos_feat, cfg, s1_weekly=s1_weekly, n_trials_local=n_loc, n_trials_stack=n_stk)
    print(oos.metrics)
    write_tearsheet_pdf(
        tearsheet_path("H-011", str(stack["VOL_STAR"]), version=VERSION, root=ROOT),
        oos.returns,
        title="H-011 sealed OOS",
        n_trials_local=n_loc,
        n_trials_stack=n_stk,
    )
else:
    print("Sealed holdout gated. Set RUN_SEALED=True after freezing VOL_STAR.")
```
